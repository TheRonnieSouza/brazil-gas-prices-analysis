```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv('./datasets/GasPricesinBrazil_2004-2019.csv', sep=';')
data
```

```python
data.head(5)
```

```python
data.reset_index(drop=True, inplace=True)
```

```python
data.shape
```

```python
data.info()
```

```python
data.drop(labels='Unnamed: 0', axis=1, inplace=True)
```

```python
data.info()
```

```python
data.isnull().sum()
```

```python
data = data.applymap(lambda x: x.strip() if isinstance(x, str) else x)
data = data.replace(r'^\s*$', np.nan, regex=True)
data.isnull().sum()
```

```python
data.info()
```

```python
data_pre = data.copy()
```

```python
data_pre['DATA INICIAL'] = pd.to_datetime(data_pre['DATA INICIAL'])
data_pre['DATA FINAL'] = pd.to_datetime(data_pre['DATA FINAL'])
```

```python
data_pre.info()
```

```python
#convertendo atributos/colunas para numeric, caso não seja adiciona not a number (nan)
for atributo in ['MARGEM MÉDIA REVENDA', 'PREÇO MÉDIO DISTRIBUIÇÃO','DESVIO PADRÃO DISTRIBUIÇÃO','PREÇO MÍNIMO DISTRIBUIÇÃO','PREÇO MÁXIMO DISTRIBUIÇÃO','COEF DE VARIAÇÃO DISTRIBUIÇÃO']:
    data_pre[atributo] = pd.to_numeric(data_pre[atributo], errors='coerce')
```

```python
data_pre.info()
```

```python
mask_is_null = data_pre['PREÇO MÉDIO DISTRIBUIÇÃO'].isnull()
data_pre[mask_is_null]
```
